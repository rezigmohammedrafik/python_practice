```python
teams=[
{"team": "Barcelona", "goals": 3},
{"team": "Real Madrid", "goals":2},
{"team": "Atletico Madrid", "goals":1},
]
base=0
for t in teams:
    base=base+t["goals"]

averagegoals=base/len(teams)   

winners = [t["team"] for t in teams if t["goals"] >=2]
print(winners)

print(f"Total goals = {base}")
print(f"Average goals = {averagegoals}")
```
