```python
import pandas as pd
teams = [
    {"team": "Barcelona", "goals": 3},
    {"team": "Real Madrid", "goals": 2},
    {"team": "Atletico Madrid", "goals": 1},
]
small=pd.DataFrame(teams)
print(small["goals"].sum())
print(small["goals"].mean()) #mean is for the average of the column
print(small[small["goals"]==1]) #this replaces the need of using the for loop 
print(small["team"])
```
